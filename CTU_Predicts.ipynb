# FSPA Predicts Cookbook

Describes how to use the close_the_u library to store and query predict data. 

## Overview

FSPA provides a web service called the (Close the U) State Data Store, which is a repository for data used in analysis of the spacecraft, that is not otherwise provided by MCS services. This includes 'predicts' / 'predicted telemetry channels, which might be generated by user tools, or may be generated in another venue e.g. WSTS or Testbed.  

The core user functions described below include:
- Creating and storing predict data
- Querying and using predict data
- Deleting and replacing predict data

Note if this notebook is run as a demo, you may want to run the 'Cleanup' step first for a clean state.

## Create Predict


This example shows...


```python


# Generate new predict records

# The following includes all the fields that are required to make records.  A number of additional fields may also have values specified
new_records = []

record1 = {}
record1["name"] = "TEST_PREDICT_1"
record1["value"] = 1
record1["scet"] = "2024-137T01:01:01.000"
record1["type"] = "predict"
record1["valueType"] = "PREDICTED"
record1["collectionName"] = COLLECTION_NAME
new_records.append(record1)

record2 = {}
record2["name"] = "TEST_PREDICT_1"
record2["value"] = 2
record2["scet"] = "2024-137T01:02:01.000"
record2["type"] = "predict"
record2["valueType"] = "PREDICTED"
record2["collectionName"] = COLLECTION_NAME
new_records.append(record1)

response = state_data_store.create_states(
    states=new_records,
    env='dev'
)

print (response)
```

### Query Predict

```python

sds_response = state_data_store.get_states_by_name(collection_name=COLLECTION_NAME, env='dev', name = 'TEST_PREDICT_1')

predict_records = sds_response["data"]["stateByName"]

print (predict_records) # (to formatted JSON)
```

### Append to Predict

```python
# Generate new records that will be appended
new_records = []

record1 = {}
record1["name"] = "TEST_PREDICT_1"
record1["value"] = 5
record1["scet"] = "2024-137T01:05:01.000"
record1["type"] = "predicted"
record1["valueType"] = "PREDICTED"
new_records.append(record1)

response = state_data_store.create_states(
    states=new_records,
    env='dev'
)

print (response)
```
