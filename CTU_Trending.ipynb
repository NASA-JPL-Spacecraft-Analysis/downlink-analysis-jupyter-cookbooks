# FPSA Trending Cookbook

Describes how to use the close_the_u library to store and query trending data. 

## Overview

FSPA provides a web service called the (Close the U) State Data Store, which is a repository for data used in analysis of the spacecraft, that is not otherwise provided by MCS services.  This includes 'trending' information such as life of mission daily/weekly etc. averages, cumulative counts and sums, etc.  It is assumed that data is generally decimated and not of the volume of high rate telemetry.  

The core user functions described below include:
- Creating and storing trend data
- Querying and using trend data
- Deleting and replacing trend data

Note if this notebook is run as a demo, you may want to run the 'Cleanup' step first for a clean state.

## Create Trend


```python
# Generate new trend records

# The following includes all the fields that are required to make records.  A number of additional fields may also have values specified
new_records = []

record1 = {}
record1["name"] = "TEST_TREND_1"
record1["value"] = 1
record1["scet"] = "2024-137T01:01:01.000"
record1["type"] = "trend"
record1["valueType"] = "DERIVED"
record1["collectionName"] = COLLECTION_NAME
new_records.append(record1)

record2 = {}
record2["name"] = "TEST_TREND_1"
record2["value"] = 2
record2["scet"] = "2024-137T01:02:01.000"
record2["type"] = "trend"
record2["valueType"] = "DERIVED"
record2["collectionName"] = COLLECTION_NAME
new_records.append(record1)

response = state_data_store.create_states(
    states=new_records,
    env='dev'
)

print (response)
```

### Query Trend

```python

sds_response = state_data_store.get_states_by_name(collection_name=COLLECTION_NAME, env='dev', name = 'TEST_TREND_1')

trend_records = sds_response["data"]["stateByName"]

print (trend_records) # (to formatted JSON)
```
